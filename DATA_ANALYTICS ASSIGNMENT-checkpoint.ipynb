```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

#Defining the  X and Y variables
X = data['Years of Experience'].values.reshape(-1, 1)
Y = data['Salary'].values

#Training sizes from 30% to 80% (step = 15)
train_sizes = np.arange(30, 85, 15)

#Lists to store MSE values
train_mse_list = []
test_mse_list = []

#For loop through each training size
for size in train_sizes:
    test_size = 1 - size / 100.0  # Convert to test size for sklearn
    
    #Splitting the data
    X_train, X_test, Y_train, Y_test = train_test_split(X, Y, test_size=test_size, random_state=42)
    
    #Training the model set
    model = LinearRegression()
    model.fit(X_train, Y_train)
    
    #Predict and calculate MSE for training and testing sets
    Y_train_pred = model.predict(X_train)
    Y_test_pred = model.predict(X_test)
    
    train_mse = mean_squared_error(Y_train, Y_train_pred)
    test_mse = mean_squared_error(Y_test, Y_test_pred)
    
    train_mse_list.append(train_mse)
    test_mse_list.append(test_mse)

#Plotting the MSEs
plt.figure(figsize=(10, 6))
plt.plot(train_sizes, train_mse_list, marker='o', label='Training MSE', color='blue')
plt.plot(train_sizes, test_mse_list, marker='o', label='Testing MSE', color='red')
plt.title('MSE vs Training Set Size Cross Validation')
plt.xlabel('Training Set Size (%)')
plt.ylabel('Mean Squared Error (MSE)')
plt.legend()
plt.grid(True)
plt.show()
```

Answer----

The cross-validation results suggest that the model is overfitting:
The model fits the training data quite well, as evidenced by the constantly low training MSE.Nevertheless, for all training sizes, the testing MSE is still much greater than the training MSE.The model may be capturing noise or extremely specialized patterns in the training data that do not translate well to fresh, unknown data, as indicated by the widening gap between the training and testing MSEs.Overfitting is evident when the model performs well on the training set but poorly on the test set.

```python
import pandas as pd

# Loading the dataset 
df = pd.read_csv('Social_Network_Ads.csv')

# Selecting the needed columns 
df = df[['Age', 'EstimatedSalary', 'Purchased']]

# Display a few columns of the dataset 
print(df.tail())
```

```python
#Dispaly the summary Statistic
print("Summary Statistics:")
print(df.describe())
```

```python
#checking for missing values befroe repalcement 
print("Missing values before replacement:")
print(df.isnull().sum())

#replacing missing values
df.fillna(df.mean(numeric_only=True), inplace=True)

#checking for missing values after replacement 
print("\nMissing values after replacement:")
print(df.isnull().sum())
```

2.4

```python
#checking for duplications
duplicate_count = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_count}")

#Dropping the  duplication and keep first occurance
df.drop_duplicates(keep='first', inplace=True)
#Confimation that duplications are removed 
print(f"Shape after removing duplicates: {df.shape}")
```

```python
import matplotlib.pyplot as plt
 

# Setting the colour of purchased values
colors = df['Purchased'].map({0: 'green', 1: 'red'})

# Creating the scatter plot
plt.figure(figsize=(8, 6))
plt.scatter(df['Age'], df['EstimatedSalary'], c=colors, alpha=0.7, edgecolors='k')

plt.title('Age vs Estimated Salary (Color by Purchased)')#name of the scatter point
plt.xlabel('Age')# Name of the "X" lable
plt.ylabel('Estimated Salary')# name of the "y" lable
plt.grid(True)
plt.show()
```

```python
#Importing all the nessacry libraries  
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier

#labling the "X" and "Y"
X = df[['Age', 'EstimatedSalary']]
y = df['Purchased']

#Splitting the test and train data set 
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

#Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

#Training k-NN classifers
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train_scaled, y_train)

print("Model trained successfully with k = 3")

accuracy = knn.score(X_test_scaled, y_test)
```

2.7

```python
# Initialize new client
new_client = pd.DataFrame([[37, 79000]], columns=['Age', 'EstimatedSalary'])

# Scaling the new client data using the  scaler
new_client_scaled = scaler.transform(new_client)

# Predict the label using the trained k-NN model
prediction = knn.predict(new_client_scaled)

# Output the prediction 
print(f"Prediction for client with Age = 37 and Salary = 79,000: {'Purchased' if prediction[0] == 1 else 'Not Purchased'}")

# 1 means "PURCHASED" 2 means "NOT PURCHASED"
```

Model evaluation
2.8

```python
#Importing all libraries from the sklearn.metrics
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score

# Making predictions on the test dataset
y_pred = knn.predict(X_test_scaled)

# Compute Accuracy
accuracy = accuracy_score(y_test, y_pred)

# Compute Precision
precision = precision_score(y_test, y_pred)

# Compute Recall
recall = recall_score(y_test, y_pred)

# Compute F1-Score
f1 = f1_score(y_test, y_pred)

# Printing the evaluation metrics of the computations
print(f"Accuracy: {accuracy:.2f}")
print(f"Precision: {precision:.2f}")
print(f"Recall: {recall:.2f}")
print(f"F1-Score: {f1:.2f}")
```

2.9

```python
from sklearn.metrics import roc_curve, auc


y_prob = knn.predict_proba(X_test_scaled)[:, 1]  # Get probabilities for class 1 

# Compute ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_prob)

# Compute AUC
roc_auc = auc(fpr, tpr)

# Plotting the ROC curve
plt.figure(figsize=(10, 8))
plt.plot(fpr, tpr, color='green', label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.plot([0, 1], [0, 1], color='red', linestyle='--')  # Diagonal line representing random chance
plt.title('Receiver Operating Characteristic Curve')
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.legend(loc='lower right')
plt.grid(True)
plt.show()

# Print the AUC score
print(f"AUC (Area Under the Curve): {roc_auc:.2f}")
```
